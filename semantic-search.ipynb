# Semantic Book Search Using Vector Embeddings

This notebook implements a semantic search system for books using vector embeddings and a vector database. we will:

1. **Load and prepare** the book dataset
2. **Export book descriptions** to a text file for processing
3. **Create document embeddings** using free HuggingFace embedding models (no API costs!)
4. **Build a vector database** using ChromaDB for efficient similarity search
5. **Implement semantic search** functionality to find books based on natural language queries
6. **Create a recommendation function** that retrieves relevant books


```python
# Import required libraries for document processing and vector storage
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import CharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_chroma import Chroma
import pandas as pd

print("Libraries imported successfully")
print("Using free HuggingFace embeddings - no API keys required!")
```

## 1. Loading the Book Dataset

we will load the cleaned book dataset that contains the indexed content (ISBN + description) for each book.


```python
# Load the book dataset
df_books = pd.read_csv("books_cleaned.csv")

print(f"Dataset loaded: {len(df_books)} books")
print(f"Columns: {list(df_books.columns)}")
print("\nSample data:")
df_books.head()
```

## 2. Preparing Text Data for Vectorization

We need to export the indexed content (ISBN + description) to a text file so we can process it with LangChain's document loaders. Each line will represent one book's content.


```python
# Export indexed content to a text file (one book per line)
text_output_file = "book_descriptions.txt"

# Write each book's indexed content to a separate line
df_books["indexed_content"].to_csv(
    text_output_file,
    sep="\n",
    index=False,
    header=False
)

print(f"Book descriptions exported to: {text_output_file}")
print(f"Total lines (books): {len(df_books)}")

# Display sample of what was exported
print("\nSample content (first 3 lines):")
with open(text_output_file, 'r') as f:
    for i, line in enumerate(f):
        if i < 3:
            print(f"Line {i+1}: {line[:150]}...")
        else:
            break
```

## 3. Loading and Splitting Documents

we will use LangChain's TextLoader to load the text file and split it into documents. Since each line represents one book, we will split by newline characters.


```python
# Load the text file as documents
raw_docs = TextLoader(text_output_file).load()

# Split documents by newline (each line is one book)
text_splitter = CharacterTextSplitter(
    chunk_size=0,      # No chunking - each line is a separate document
    chunk_overlap=0,   # No overlap needed
    separator="\n"     # Split by newline character
)

book_documents = text_splitter.split_documents(raw_docs)

print(f"Total documents created: {len(book_documents)}")
print(f"\nSample document:")
print(f"Content preview: {book_documents[0].page_content[:200]}...")
print(f"Metadata: {book_documents[0].metadata}")
```

## 4. Creating Vector Database with Embeddings

we will create a ChromaDB vector database that stores document embeddings. This allows us to perform fast similarity searches using vector distance calculations.

**Using Free HuggingFace Embeddings:**
- Model: `BAAI/bge-large-en-v1.5` - High quality, best performance, 1024 dims
- Runs locally (no API calls, no costs)
- First time: Model will be downloaded from HuggingFace (one-time download)
- Subsequent runs: Uses cached model

**Important:** The database is persisted to disk (`./chroma_db`), so embeddings are only created once. On subsequent runs, the existing database will be loaded, saving time.


```python
# Initialize HuggingFace embeddings model
# Using 'BAAI/bge-large-en-v1.5' - state-of-the-art embedding model
# This model provides excellent quality for semantic search
# Alternative options:
#   - 'sentence-transformers/all-MiniLM-L6-v2' (faster, smaller)
#   - 'sentence-transformers/all-mpnet-base-v2' (better quality, slower)
embedding_model = HuggingFaceEmbeddings(
    model_name="BAAI/bge-large-en-v1.5",
    model_kwargs={'device': 'mps'},  # Use 'cuda' if you have GPU
    encode_kwargs={'normalize_embeddings': True}  # Normalize for better similarity search
)

print("HuggingFace embedding model initialized")
print("Model: BAAI/bge-large-en-v1.5 (free, runs locally, no API calls)")
print("Note: This is a state-of-the-art model with excellent semantic search quality")

# Define directory for persisting the vector database
persist_directory = "./chroma_db"

import os

# Check if vector database already exists
if os.path.exists(persist_directory) and os.listdir(persist_directory):
    print("\nLoading existing vector database from disk...")
    print("(Embeddings already created - no computation needed)")
    
    # Load existing database
    vector_db = Chroma(
        persist_directory=persist_directory,
        embedding_function=embedding_model
    )
    
    print(f"Vector database loaded successfully!")
    print(f"Total documents in database: {vector_db._collection.count()}")
else:
    print("\nCreating new vector database...")
    print("This will generate embeddings for all book descriptions...")
    print("(This may take a few minutes - running locally, no API calls)")
    
    # Create new database and persist to disk
    vector_db = Chroma.from_documents(
        documents=book_documents,
        embedding=embedding_model,
        persist_directory=persist_directory
    )
    
    print(f"\nVector database created and saved to: {persist_directory}")
    print(f"Total documents indexed: {len(book_documents)}")
    print("Embeddings are now stored locally and will be reused in future runs!")
```

## 5. Testing Semantic Search

Let's test the semantic search functionality with a sample query. The system will find books whose descriptions are semantically similar to the query.


```python
# Test semantic search with a sample query
test_query = "A book to teach children about nature"
num_results = 10

print(f"Query: '{test_query}'")
print(f"Retrieving top {num_results} most similar books...\n")

similar_books = vector_db.similarity_search(test_query, k=num_results)

print(f"Found {len(similar_books)} results:\n")
for idx, doc in enumerate(similar_books, 1):
    # Extract ISBN from the document content (first token)
    content_parts = doc.page_content.split()
    isbn = content_parts[0] if content_parts else "Unknown"
    description_preview = " ".join(content_parts[1:])[:150]
    
    print(f"Result {idx}:")
    print(f"  ISBN: {isbn}")
    print(f"  Description preview: {description_preview}...")
    print()
```

```python
# Retrieve the full book information for the top result
if similar_books:
    top_result = similar_books[0]
    top_isbn = int(top_result.page_content.strip('"').split()[0])
    
    matching_book = df_books[df_books["isbn13"] == top_isbn]
    
    if not matching_book.empty:
        print("Top matching book details:")
        print(matching_book[['title', 'authors', 'categories', 'description']].to_string())
    else:
        print(f"Book with ISBN {top_isbn} not found in dataset")
```

## 6. Creating Semantic Recommendation Function

we will create a reusable function that takes a natural language query and returns relevant book recommendations based on semantic similarity.


```python
def get_semantic_book_recommendations(
        search_query: str,
        num_recommendations: int = 10
) -> pd.DataFrame:
    """
    Retrieve book recommendations based on semantic similarity to a query.
    
    Args:
        search_query: Natural language query describing desired book
        num_recommendations: Number of recommendations to return (default: 10)
        
    Returns:
        DataFrame containing recommended books with full metadata
    """
    # Perform similarity search in vector database
    # Retrieve more results than requested to ensure we have enough matches
    search_results = vector_db.similarity_search(search_query, k=min(50, num_recommendations * 5))
    
    # Extract ISBNs from search results
    recommended_isbns = []
    for result in search_results:
        # Extract ISBN from document content (first token)
        content_text = result.page_content.strip('"')
        isbn_str = content_text.split()[0]
        try:
            isbn_int = int(isbn_str)
            recommended_isbns.append(isbn_int)
        except ValueError:
            # Skip if ISBN cannot be converted to int
            continue
    
    # Filter books dataframe to only include recommended books
    recommended_books = df_books[df_books["isbn13"].isin(recommended_isbns)]
    
    # Limit to requested number of recommendations
    return recommended_books.head(num_recommendations)

# Test the function
print("Testing recommendation function...")
recommendations = get_semantic_book_recommendations("A book to teach children about nature", num_recommendations=10)
print(f"\nRetrieved {len(recommendations)} recommendations")
print("\nRecommended books:")
print(recommendations[['title', 'authors', 'categories']].to_string())
```

## 7. Additional Search Examples

Let's test the semantic search with different types of queries to demonstrate its versatility.


```python
# Example query 1: Fiction recommendation
query1 = "A thrilling mystery novel with unexpected twists"
results1 = get_semantic_book_recommendations(query1, num_recommendations=5)

print(f"Query: '{query1}'")
print(f"\nTop 5 recommendations:")
for idx, row in results1.iterrows():
    print(f"\n{idx+1}. {row['title']}")
    print(f"   Author: {row['authors']}")
    print(f"   Category: {row.get('categories', 'N/A')}")
    print(f"   Description: {str(row['description'])[:200]}...")
```

```python
# Example query 2: Nonfiction recommendation
query2 = "A comprehensive guide to world history"
results2 = get_semantic_book_recommendations(query2, num_recommendations=5)

print(f"Query: '{query2}'")
print(f"\nTop 5 recommendations:")
for idx, row in results2.iterrows():
    print(f"\n{idx+1}. {row['title']}")
    print(f"   Author: {row['authors']}")
    print(f"   Category: {row.get('categories', 'N/A')}")
    print(f"   Description: {str(row['description'])[:200]}...")
```
